### Filter our data using a lambda function

For that we will:
- create a column that contains a Boolean, `True` or `False`, based on a condition
- filter down our data based on that condition

We can do this using `functions` like the ones we've learned about in early Python lessons that we declare using `def`. We can also do this using`lambda function` which are functions written in one line!

Below we do the same thing we did earlier to reduce our dataframe to only `"per Annum"` workers but with a bit more Python.  

````python
data["is_perannum?"] = data["Pay Basis"].apply(lambda x: "per annum" in str(x).lower())
data["is_perannum?"]
````

````python
reduced_data = data[data["is_perannum?"] == True]
````

````python
data_by_agency = data["Agency Name"].value_counts().reset_index()
data_by_agency.head()

# len(data)
````

````python
top_20 = data_by_agency.sort_values(by="count", ascending=False).head(20)
print(top_20)
````

````python
top_20.plot(x="Agency Name", y="count", kind="barh", color="green")
````

````python
median_salary_by_agency = data.groupby(
    ['Agency Name']
)['Base Salary'].median(

).reset_index(

).sort_values(
    by='Base Salary',
    ascending=False
)

median_salary_by_agency.head()

top_10_salaries = median_salary_by_agency.sort_values(by='Base Salary',ascending=False).head(10)
````

````python
top_10_salaries.plot(x="Agency Name", y="Base Salary", kind="barh", color="#ddd")
````
