### Data aggregation analysis of campaign finance data in NYC

In this notebook, we will analyze and aggregate raw campaign finance data using the [`.groupby()` function](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html) in pandas. 

We will:
- import dependencies
- open and explore the data
- aggregate the data in different ways

```python
import pandas as pd
```

```python
dtypes = {
    "RECIPID":"str",
    "INTZIP":"str",
    "INT_C_CODE":"str"
    
}

date_columns =["Fiscal Year", "Agency Start Date"]

salary_data = pd.read_csv(
    "../data/Citywide_Payroll_Data__Fiscal_Year__20240616.csv", 
    dtype=dtypes, 
    parse_dates=date_columns)

print(
    len(salary_data),
    salary_data.columns
)
```

```python
salary_data.sort_values(
    by="Base Salary",
    ascending=False
)
```

### Aggregate data in different ways 

- recipient name with the most number of donations
- recipient name with the most number of donations and the highest total amounts of donations
- occupation that occur the most in the data by city

```python
salary_data["Title Description"].value_counts()
```

```python
# This code takes your salary_data DataFrame, groups employees by their 
# job title, calculates the median base salary for each title, 
# and sorts the results from highest to lowest median salary.

salary_data.groupby( ["Title Description"]
    )["Base Salary"].median(
    
    ).reset_index(
    
    ).sort_values(
        by="Base Salary",
        ascending=False
)
```

```python
#This code groups employees by job title, calculates the median base salary, 
# median overtime pay, and median other pay for each title, and then sorts the 
# results from highest to lowest median base salary.

#It's similar to your previous example, but instead of calculating the median 
# for just one column, it calculates medians for three different columns at once.


salary_data.groupby(
        ["Title Description"]
    ).agg(
        {
            "Base Salary":"median",
            "Total OT Paid":"median",
            "Total Other Pay":"median"
        }
    ).sort_values(
        by="Base Salary",
        ascending=False
)
```

```python

salary_data.groupby(
        ["Agency Name", "Work Location Borough"]
    ).agg(
        {
            "Base Salary":"median"
        }
    ).sort_values(
        by="Base Salary",
        ascending=False
)
```
