```python
data.groupby(
    ["state"]
    )["age"].median(

    ).reset_index(

    ).sort_values(
        by="age",
        ascending=True
    )
```

```python
data.columns
```

```python
data.groupby(
    ["state", "race"]
).agg(
    {
        "age":"mean"
    }
).sort_values(
    by="state",
    ascending=True
)
```

```python
# Figure out which state had the most shootings

data["state"].value_counts().sort_values(ascending=True)
```

```python
# What's the average age of the victims in the dataset?

data["age"].describe()
```

```python
# Figure out the average age of unarmed victims per state

data.groupby(
    ["state"]
).agg(
    {"age":"mean"}
).reset_index(

).sort_values(
    by="age",
    ascending=True
)
```

```python
# Figure out how many cases were mental illness related

data["was_mental_illness_related"].value_counts()
```

```python
# ... as a percent over the total?
percent_mental_illness_related = data["was_mental_illness_related"].value_counts()[True] / len(data) * 100

print(percent_mental_illness_related)

# remember: .value_counts() counts occurrences of values
```

```python
# Group the data by city and state. Which city had the most shootings?

data.groupby(
    ["city", "state"]
).size().sort_values(ascending=False)

# size() on the other hand counts rows in a group!
```
