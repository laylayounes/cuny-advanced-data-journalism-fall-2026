```python
# Open and read the html file (remember that your file path may be different!!!)
with open("../data/project_gutenberg.html", "r") as page:

    # Tell the BeautifulSoup library to interpret the file
    soup = BeautifulSoup(page, "html.parser")

    # Find the first h1 on the page. Print the text stored within it.
    print(soup.find("h1").get_text())

    # There are three follow links on the page, one for Facebook, one for Bluesky, and one for Mastodon. 
    # What is the class associated with these follow links? Answer here: navlink
    # What html element are these follow links stored in? Answer here: <li>

    # Use the function soup.find_all(...) to grab the html associated with the three follow links and store them in an array.
    nav_links = soup.find_all("li",{"class":"navlink"})


    # first_link = nav_links[0]
    # print(first_link.find("a").get('href'))

    # Loop through the array, printing the text for each follow link.
    # for nav_link in nav_links:
    #     print(nav_link.prettify())
        # print(nav_link.get_text())

    # Let's take it a step further. Paste your loop again below. In each iteration of the loop, 
    # grab the first <span></span> that you find 
    # in the html code and print out its text.
    for nav_link in nav_links:
        # first_span = nav_link.find("span", {"class":"title"})
        print(nav_link.find("a").find("span",{"class":"title"}).get_text())
        # print(first_span.get_text())
        



```
