```python
import csv 
import time
from bs4 import BeautifulSoup
import requests
```

### open the web site with requests

```python

url = "https://en.wikipedia.org/wiki/Category:Sufi_mystics"

# Your identification
headers = {"user-agent" : "Mozilla/5.0 (Windows NT 6.1) AppleWebKit/537.36(KHTML, like Gecko) Chrome/41.0.2228.0 Safari/537.36;",
"from": "Lylla Younes lylla.younes@journalism.cuny.edu"}

# Send a request to Wikipedia and pull down the Sufi mystics page
page = requests.get(url, headers=headers)

# Isolate the html from the webpage
page_content = page.content

```

### parse the page through the BeautifulSoup library

```python
# Ask BeautifulSoup to parse the html
soup = BeautifulSoup(page_content, "html.parser")

# mw-category --> mw-category-group which contains <h3> and an <ul> with the 
# list items <li> that contain my information

data_container = (soup.find_all("div", {"class":"mw-category"}))[1]
print(data_container.prettify())
```

```python
groupings = data_container.find_all("div", {"class":"mw-category-group"})

sufi_mystics = []

for group in groupings:
    # print(group.prettify())
    category_name = group.find("h3").get_text()
    # print(category_name)
    list_items = group.find_all("li")

    
    for list_item in list_items:
        # print(list_item.prettify())
        name = list_item.text
        url = "https://en.wikipedia.org/" + list_item.find("a").get('href')
        d = {"name":name, "url":url, "category":category_name}

        sufi_mystics.append(d)

print(sufi_mystics)
```
