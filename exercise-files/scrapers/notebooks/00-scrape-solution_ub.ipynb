```python
# Now let's take a look at ALL cities in the world!

url = "https://www.uber.com/us/en/r/cities/"

# Set the headers
headers = {
"User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 "
        "(KHTML, like Gecko) Chrome/124.0 Safari/537.36"
    )
}

# Sends an HTTP GET request to the specified URL with custom headers and a maximum 15-second wait time
response = requests.get(url, headers=headers, timeout=15)

# Raises an error automatically if the website returned a failure status code
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

cities = []

country_links = []
divs = soup.find_all("div", {"class": "css-diKeRO"})
for div in divs:
    links = div.find_all("a")
    print(links[0]["href"])
    country_links.append(links[0]["href"])

# print(country_links)
```

```python
city_links = []
for country_link in country_links:

    print(country_link)
    response = requests.get(country_link, headers=headers, timeout=15)
    soup = BeautifulSoup(response.text, "html.parser")

    divs = soup.find_all("div", {"class": "css-diKeRO"})
    for div in divs:
        links = div.find_all("a")
        print(links[0]["href"])
        city_links.append(links[0].text)


# print(city_links)
```
