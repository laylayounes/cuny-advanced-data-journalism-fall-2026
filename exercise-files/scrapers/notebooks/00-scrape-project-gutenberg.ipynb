### Let's scrape Project Gutenberg

https://www.gutenberg.org/ebooks/search/?sort_order=release_date


```python
import csv
from bs4 import BeautifulSoup
```

```python

book_list = []

with open("../data/project_gutenberg.html", "r") as page:
    soup = BeautifulSoup(page,  "html.parser")

    books = soup.find_all("li", {"class":"booklink"})

    for list_item in books:

        title  = list_item.find("span", {"class", "title"})
        author = list_item.find("span", {"class", "subtitle"})
        date   = list_item.find("span", {"class", "extra"})

        # print(title.text)
        # print(author.text)
        # print(date.text)

        info = {"title":title.text, "author":author.text, "date":date.text}
        book_list.append(info)

fieldnames = ["title", "author", "date"]
with open("../output/gutenberg_books.csv", mode="w", newline="", encoding="utf-8") as csvfile:

    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    # Write the header row
    writer.writeheader()
    
    # Write all rows at once
    writer.writerows(book_list)


    
    
    
```
