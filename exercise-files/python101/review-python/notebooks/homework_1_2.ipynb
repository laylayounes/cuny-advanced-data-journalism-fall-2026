```python
# Import the pycountry library.
import pycountry
```

```python
# Check how many countries are in the library.
len(pycountry.countries)
```

```python
# Loop through every country in the pycountries library. 
# The data type for each country is a tuple. 
# Let's convert and display each country as a dictionary. 

for country in pycountry.countries:
    print(dict(country))
```

Now your turn. Loop through the countries and print every flag.

```python
for country in pycountry.countries:
    # print(dict(country)['flag'])
    print(country.flag)
```

Let's take it a step further. Add each flag to a list called flags. 

```python
flags = []

for country in pycountry.countries:
    flags.append(country.flag)

print(flags)
```

Now use a loop to populate an empty dictionary. 
The keys should be country names, and the values should be flags.

```python
countries = {}

for country in pycountry.countries:
    countries[country.name] = country.flag

print(countries)
```

```python
print(countries.keys())
print(countries.values())

for key in countries:
    print(key)
```

```python
# Now let's do some file writing. Create a new csv file and populate it using the dictionary you just built.

import csv
```

```python
with  open("../output/country_flags.csv", "w") as csvfile:
    
    writer = csv.writer(csvfile)
    writer.writerow(["country","flag"])

    for key in countries.keys():
        writer.writerow([key,countries[key]])

```
